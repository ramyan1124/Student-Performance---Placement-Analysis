```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
students=pd.read_csv("/content/indian_student_placement_data.csv")
students.head()
```

```python
students.info()
```

```python
students.describe()
```

```python
students.duplicated().sum()
```

```python
students = students.drop(columns=['Email'])

students.head()
```

```python
students.columns
```

```python
students['Placement Date'] = pd.to_datetime(students['Placement Date'])
students.dtypes
```

```python
students.describe()
```

```python
students['Course'].value_counts()
```

## 📊 Course-wise Student Distribution

```python
course_counts = students['Course'].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(course_counts.index, course_counts.values)

plt.title('Students by Course')
plt.xlabel('Course')
plt.ylabel('Number of Students')

plt.show()
```

## 🏫 Branch-wise Student Distribution

```python
branch_counts = students['Branch'].value_counts()

plt.figure(figsize=(10, 5))
plt.bar(branch_counts.index, branch_counts.values)

plt.title('Students by Branch')
plt.xlabel('Branch')
plt.ylabel('Number of Students')

plt.xticks(rotation=45)
plt.show()
```

## 🏢 Company-wise Placement Distribution

```python
company_counts = students['Company'].value_counts()

plt.figure(figsize=(10, 5))
plt.bar(company_counts.index, company_counts.values)

plt.title('Students Placed by Company')
plt.xlabel('Company')
plt.ylabel('Number of Students')

plt.xticks(rotation=45)
plt.show()
```

## 💰 Salary Distribution

```python
plt.figure(figsize=(10, 5))

plt.hist(students['Salary (INR)'], bins=20)

plt.title('Salary Distribution of Placed Students')
plt.xlabel('Salary (INR)')
plt.ylabel('Number of Students')

plt.show()
```

## 🏢💰 Average Salary by Company

```python
avg_salary_company = students.groupby('Company')['Salary (INR)'].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
plt.bar(avg_salary_company.index, avg_salary_company.values)

plt.title('Average Salary by Company')
plt.xlabel('Company')
plt.ylabel('Average Salary (INR)')

plt.xticks(rotation=45)
plt.show()
```

## 💼 Job Role Distribution

```python
job_role_counts = students['Job Role'].value_counts()

plt.figure(figsize=(10, 5))
plt.bar(job_role_counts.index, job_role_counts.values)

plt.title('Students by Job Role')
plt.xlabel('Job Role')
plt.ylabel('Number of Students')

plt.xticks(rotation=45)
plt.show()
```

## 💼💰 Average Salary by Job Role

```python
avg_salary_role = students.groupby('Job Role')['Salary (INR)'].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
plt.bar(avg_salary_role.index, avg_salary_role.values)

plt.title('Average Salary by Job Role')
plt.xlabel('Job Role')
plt.ylabel('Average Salary (INR)')

plt.xticks(rotation=45)
plt.show()
```

## 📍 Location-wise Placement Distribution

```python
location_counts = students['Location'].value_counts()

plt.figure(figsize=(10, 5))
plt.bar(location_counts.index, location_counts.values)

plt.title('Students Placed by Location')
plt.xlabel('Location')
plt.ylabel('Number of Students')

plt.xticks(rotation=45)
plt.show()
```

## 📍💰 Average Salary by Location

```python
avg_salary_location = students.groupby('Location')['Salary (INR)'].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
plt.bar(avg_salary_location.index, avg_salary_location.values)

plt.title('Average Salary by Location')
plt.xlabel('Location')
plt.ylabel('Average Salary (INR)')

plt.xticks(rotation=45)
plt.show()
```

## 📅 Placement Trend by Year

```python
students['Placement Year'] = students['Placement Date'].dt.year

placement_year_counts = students['Placement Year'].value_counts().sort_index()

plt.figure(figsize=(8, 5))
plt.plot(
    placement_year_counts.index,
    placement_year_counts.values,
    marker='o')

plt.title('Placement Trend by Year')
plt.xlabel('Placement Year')
plt.ylabel('Number of Students')

plt.xticks(placement_year_counts.index)
plt.show()
```

## 🎓💰 Average Salary by Course

```python
avg_salary_course = (
    students.groupby('Course')['Salary (INR)']
    .mean()
    .sort_values(ascending=False))

plt.figure(figsize=(8, 5))
plt.bar(avg_salary_course.index, avg_salary_course.values)

plt.title('Average Salary by Course')
plt.xlabel('Course')
plt.ylabel('Average Salary (INR)')

plt.show()
```

## 🎓💰 Average Salary by Branch

```python
avg_salary_branch = (
    students.groupby('Branch')['Salary (INR)']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 5))
plt.bar(avg_salary_branch.index, avg_salary_branch.values)

plt.title('Average Salary by Branch')
plt.xlabel('Branch')
plt.ylabel('Average Salary (INR)')

plt.xticks(rotation=45)
plt.show()
```

## 🎓💰 Salary vs Course

```python
plt.figure(figsize=(10, 5))

sns.boxplot(
    data=students,
    x='Course',
    y='Salary (INR)'
)

plt.title('Salary Distribution by Course')
plt.xlabel('Course')
plt.ylabel('Salary (INR)')

plt.show()
```

## 🎓💼 Course vs Job Role

```python
course_job = pd.crosstab(
    students['Course'],
    students['Job Role']
)

plt.figure(figsize=(12, 6))

sns.heatmap(
    course_job,
    annot=True,
    fmt='d',
    cmap='Blues'
)

plt.title('Course vs Job Role Distribution')
plt.xlabel('Job Role')
plt.ylabel('Course')

plt.show()
```

## 📊 Correlation Analysis

```python
correlation = students[
    ['Graduation Year', 'Salary (INR)', 'Placement Year']
].corr()

plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Matrix')

plt.show()
```

## 🏆 Top 10 Highest Salary Students

```python
top_10_salary = students.nlargest(10, 'Salary (INR)')

top_10_salary[
    ['Name', 'Course', 'Branch', 'Company', 'Job Role', 'Salary (INR)']
]
```

# 📌 Key Insights

```python
print("Total Students:", len(students))
print("Average Salary: ₹", round(students['Salary (INR)'].mean()))
print("Median Salary: ₹", round(students['Salary (INR)'].median()))

print("\nMost Common Course:", students['Course'].mode()[0])
print("Most Common Branch:", students['Branch'].mode()[0])
print("Most Common Company:", students['Company'].mode()[0])
print("Most Common Job Role:", students['Job Role'].mode()[0])

print("\nHighest Salary: ₹", students['Salary (INR)'].max())
print("Lowest Salary: ₹", students['Salary (INR)'].min())

print("\nMost Common Placement Location:",
      students['Location'].mode()[0])
```

### 🔍 Key Findings

- The dataset contains **1,000 placed students** across different courses, branches, companies, and job roles.
- The **average salary is ₹7.40 lakh**, while the median salary is approximately **₹7.46 lakh**.
- **BA** is the most represented course in the dataset.
- **EEE** is the most represented branch.
- **IBM** appears as the most common recruiting company.
- **Support Engineer** is the most common job role.
- Salaries range from approximately **₹3.01 lakh to ₹11.98 lakh**.
- **Bangalore** is the most common placement location in the dataset.

* The analysis provides a structured view of placement outcomes and can help understand how academic background, job roles, companies, and locations relate to salary patterns within the dataset.
